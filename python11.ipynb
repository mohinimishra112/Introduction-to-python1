```python
class Person:
    # Constructor (__init__ method) 
    def __init__(self, name, age, gender):
        self.name = name       
        self.age = age        
        self.gender = gender  

    
    def speak(self):
        return f"Hi, my name is {self.name} and I am {self.age} years old."

    
    def walk(self):
        return f"{self.name} is walking."
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name  
        self.age = age    

    
    def introduce(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."


    def have_birthday(self):
        self.age += 1  
        return f"Happy Birthday! You are now {self.age} years old."
```

```python
person_1 = Person("Alice", 30)
```

```python
print(person_1.introduce()) 
```

```python
print(person_1.have_birthday())  
```

```python
class Car:
    
    wheels = 4

    def __init__(self, make, model):
        self.make = make  
        self.model = model  

    # Class method
    @classmethod
    def set_wheels(cls, number):
        cls.wheels = number  

    def car_info(self):
        return f"{self.make} {self.model} has {Car.wheels} wheels."
```

```python
car1 = Car("Toyota", "Corolla")
```

```python
car1.wheels
```

```python
car1.set_wheels(6)
```

```python
car1.wheels
```

```python
car2 = Car("Honda", "Civic")
```

```python
car2.wheels
```
