```python

def div(a, b):
    if b != 0:
        result = a / b
        print(f"The division of {a} by {b} is {result}.")
    else:
        print("Division by zero is not allowed.")
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
div(num1, num2)
```

```python

def square(x):
    result = x * x
    print(f"The square of {x} is {result}.")
num = float(input("Enter a number to find its square: "))
square(num)
```

```python
import random
numbers = [random.randint(1, 100) for _ in range(5)]
print(f"The list of random numbers is: {numbers}")
print(f"The maximum value is: {max(numbers)}")
print(f"The minimum value is: {min(numbers)}")
```
