```python


class GoldenVillas:
    def __init__(self,a,b,c):
        self.color = a
        self.is_furnish = b
        self.is_rent  = c
    
    def describe_villa(self):
        print(f"This villa is having {self.color} and it {self.is_furnish} and it is {self.is_rent} property")
        
        
```

```python
villa_1 = GoldenVillas("Blue","furnished","rented")
```

```python
villa_1.color
```

```python
villa_1.is_furnish
```

```python
villa_1.is_rent
```

```python
villa_1.describe_villa()
```

```python

class Villa:
    def __init__(self, a, b, c):
        self.color = a
        self.rooms = b
        self.size = c

    def describe_villa(self):
        return f"This villa has {self.rooms} rooms, is {self.size} sqft, and is painted {self.color}."
```

```python
villa_1 = Villa("White", 5, 2000)  
```

```python
villa_1.color
```

```python
villa_1.describe_villa()
```

```python
villa_2 = Villa("Blue", 4, 1800)   
```
